# Sparkify ETL — Exploration
This notebook documents the main ETL flow and schema at a high level.

```python
import pandas as pd
from pathlib import Path

song_files = list(Path('data/song_data').rglob('*.json')) if Path('data/song_data').exists() else []
log_files = list(Path('data/log_data').rglob('*.json')) if Path('data/log_data').exists() else []
print(f'Song files: {len(song_files)}')
print(f'Log files: {len(log_files)}')
```

```python
# Inspect one event-log file when the Sparkify dataset is available.
if log_files:
    events = pd.read_json(log_files[0], lines=True)
    display(events.head())
    print(events['page'].value_counts(dropna=False))
```

## Pipeline flow
1. Read JSON song metadata and event logs.
2. Filter streaming activity to `NextSong` events.
3. Transform timestamps into the time dimension.
4. Load users, songs, artists and songplays into PostgreSQL.
